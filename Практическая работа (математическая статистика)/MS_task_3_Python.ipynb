## Критерий Неймана–Пирсона (простая гипотеза против простой альтернативы)
##  и χ² Пирсона


В файле data_hypothesis.xlsx содержатся четыре выборки объема $n = 1000$:
- Лист **Sample1**
- Лист **Sample2**
- Лист **Sample3**

Требуется:

### Задание  1. Простая нулевая гипотеза против простой альтернативы
Для выборки **Sample1** проверить гипотезу
$$H_0: \mu = 2.4$$
против простой альтернативы
$$H_1: \mu = 2.5$$
при **известной дисперсии** $\sigma^2 = 1$.

Постройте наиболее мощный критерий (критерий Неймана–Пирсона) для уровней значимости $\alpha = 0.06$ и $\alpha = 0.01$.
- Определите критическую область (порог для выборочного среднего).
- Вычислите **ошибку первого рода** (она равна заданному $\alpha$) и **ошибку второго рода** $\beta$ (вероятность принять $H_0$, когда верна $H_1$).
- По имеющейся выборке вычислите выборочное среднее и проверьте, отвергается ли нулевая гипотеза на каждом уровне значимости.
- Также вычислите p-value (одностороннее) и сравните с $\alpha$.

### Задание  2. Критерий χ² Пирсона
Для выборок **Sample2**, **Sample3** и **Sample4** проверить гипотезу
$$H_0 : X \sim \mathcal{N}(3, 1)$$
при уровне значимости $\alpha = 0.05$.
Число интервалов $k$ — по правилу Стёрджеса: $k = \lfloor 1 + \log_2 n \rfloor$, интервалы равной длины.
- Вычислить ожидаемые и наблюдаемые частоты, при необходимости объединить интервалы с ожидаемой частотой $< 5$.
- Использовать scipy.stats.chisquare для получения статистики χ² и p-value.
- Для справки вычислить критическое значение χ².
- Сделать выводы.

```python
import numpy as np
import pandas as pd
from scipy import stats
from scipy.stats import norm, chi2, chisquare
```

###Задание №1

```python
# Читаем данные и записываем их в массив
df1 = pd.read_excel('data_hypothesis.xlsx', sheet_name='Sample1', header=None)
sample1 = df1.values.flatten()
n = len(sample1)
```

```python
H0 = 2.4
H1 = 2.5
sigma = np.sqrt(1)

alpha_values = [0.06, 0.01]
```

```python
for alpha in alpha_values:
    print(f"\n--- Уровень значимости alpha = {alpha} ---")

    # Ошибка первого рода
    print(f"  Ошибка первого рода = {alpha}")

    # Критическое значение для правостороннего критерия (H1 > H0)
    z_crit = norm.ppf(1 - alpha)
    print(f"  Критическое значение z_(1-alpha) = {z_crit:.4f}")

    # Порог для выборочного среднего
    threshold = H0 + z_crit * sigma / np.sqrt(n)
    print(f"  Критическое значение выборочного среднего = {threshold:.4f}")
    print(f"  Критическая область: выборочное среднее > {threshold:.4f}")

    # Проверка гипотезы по выборке
    x_bar = np.mean(sample1)
    reject = x_bar > threshold
    print(f"  Выборочное среднее = {x_bar:.4f}")
    print(f"  Результат: {'Отвергаем H0' if reject else 'Нет оснований отвергнуть H0'}")

    # Ошибка второго рода
    z_beta = (threshold - H1) / (sigma / np.sqrt(n))
    beta = norm.cdf(z_beta)
    power = 1 - beta
    print(f"  Ошибка второго рода = {beta:.6f}")
    print(f"  Мощность критерия = {power:.6f}")

    # Статистика критерия и p-value
    z_stat = (x_bar - H0) / (sigma / np.sqrt(n))
    p_value = 1 - norm.cdf(z_stat)
    print(f"  Статистика критерия Z = {z_stat:.4f}")
    print(f"  p-value (одностороннее) = {p_value:.6f}")
    print(f"  Сравнение: p-value {p_value:.6f} {'<' if p_value < alpha else '>='} alpha = {alpha}")
    print(f"  Вывод по p-value: {'Отвергаем H0' if p_value < alpha else 'Нет оснований отвергнуть H0'}")
```

###Задание №2

```python
df2 = pd.read_excel('data_hypothesis.xlsx', sheet_name='Sample2', header=None)
df3 = pd.read_excel('data_hypothesis.xlsx', sheet_name='Sample3', header=None)
df4 = pd.read_excel('data_hypothesis.xlsx', sheet_name='Sample4', header=None)

sample2 = df2.values.flatten()
sample3 = df3.values.flatten()
sample4 = df4.values.flatten()
```

```python
teta, sigma = 3, 1
alpha = 0.05
```

```python
def chi2_test(sample, name):
    n = len(sample)
    print(f"\n{name}: n = {n}")

    # Число интервалов по правилу Стерджеса
    k = int(1 + np.log2(n))
    print(f"  k = {k}")

    # Границы интервалов (равной длины)
    bins = np.linspace(np.min(sample), np.max(sample), k + 1)

    # Наблюдаемые частоты
    observed, _ = np.histogram(sample, bins=bins)

    # Ожидаемые частоты
    expected = []
    for i in range(len(bins)-1):
        prob = norm.cdf(bins[i + 1], teta, sigma) - norm.cdf(bins[i], teta, sigma)
        expected.append(prob * n)

    # Объединение интервалов с ожидаемой частотой < 5
    obs_merged = []
    exp_merged = []
    i = 0
    while i < len(observed):
        temp_o = observed[i]
        temp_e = expected[i]
        while temp_e < 5 and i + 1 < len(observed):
            i += 1
            temp_o += observed[i]
            temp_e += expected[i]
        obs_merged.append(temp_o)
        exp_merged.append(temp_e)
        i += 1

    # Проверка последнего интервала
    if len(exp_merged) >= 2 and exp_merged[-1] < 5:
        obs_merged[-2] += obs_merged[-1]
        exp_merged[-2] += exp_merged[-1]
        obs_merged.pop()
        exp_merged.pop()

    # Нормализация ожидаемых частот, чтобы сумма совпадала с наблюдаемыми
    sum_obs = sum(obs_merged)
    sum_exp = sum(exp_merged)
    exp_merged = [e * sum_obs / sum_exp for e in exp_merged]

    # Статистика хи-квадрата через scipy.stats.chisquare
    chi2_stat, p_value = chisquare(f_obs=obs_merged, f_exp=exp_merged, ddof=2)

    # Степени свободы
    df = len(obs_merged) - 1 - 2

    # Критическое значение (для справки)
    crit = chi2.ppf(1 - alpha, df)

    print(f"  Интервалов после объединения: {len(obs_merged)}")
    print(f"  хи-квадрат = {chi2_stat:.4f}")
    print(f"  df = {df}")
    print(f"  p-value = {p_value:.6f}")
    print(f"  Критическое значение = {crit:.4f}")
    print(f"  Вывод: {'Отвергаем H0' if p_value < alpha else 'Не отвергаем H0'}")
```

```python
chi2_test(sample2, "Sample2")
chi2_test(sample3, "Sample3")
chi2_test(sample4, "Sample4")
```
